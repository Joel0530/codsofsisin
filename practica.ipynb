```python
import string
palabra = input("escribe tu palabra")
letras= list(palabra)
mayus = string.ascii_uppercase
minus = string.ascii_lowercase
print(letras)
for i in letras:
    if i in mayus:
        print("mayuscula",i)
    elif i in minus:
        print("minuscula",i)
else:
        print("es un simbolo o un numero")
```
